```julia
using PoincareMaps
using TaylorIntegration
using TaylorSeries
using LinearAlgebra
```

```julia
function f!(dx, x, params, t)
    dx[1] = x[2]
    dx[2] = -x[1]
end
```

```julia
q0 = [1.0, 0.0]
```

```julia
params = nothing
```

```julia
function bc!(x, params, t)
    nothing
end
```

```julia
function g(dx, x, params, t)
    return (true, x[1])
end
```

```julia
t0 = 0.0
tmax = 100.0

abstol = 1e-16
order = 20

maxevents = 10
maxsteps = 1000
```

```julia
cache = PoincareMaps.init_cache_ps(
    t0,
    q0,
    maxevents,
    order,
    f!,
    params;
    parse_eqs=true,
)
```

```julia
nevents = PoincareMaps.taylorinteg_ps!(
    f!,
    bc!,
    g,
    q0,
    t0,
    tmax,
    abstol,
    cache,
    params;
    maxsteps=maxsteps,
    maxevents=maxevents,
    eventorder=1,
)
```

```julia
transpose(cache.xv[:, 1:nevents-1])
```
