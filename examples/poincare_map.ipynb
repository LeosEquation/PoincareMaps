```julia
using PoincareMaps
using TaylorIntegration
using TaylorSeries
using LinearAlgebra
```

```julia
function H(x, params, t)
    q, p = x
    return 0.5 * (q^2 + p^2)
end

@taylorize function f!(dx, x, params, t)
    dx[1] = x[2]
    dx[2] = -x[1]
end
```

```julia
function bc!(x, params, t)
    nothing
end
```

```julia
function g(dx, x, params, t)

    q = x[1]
    p = x[2]

    # Sección: q = 0
    #
    # Condición adicional:
    # solamente aceptamos cruces con p > 0.
    return (p > 0, q)
end
```

```julia
function nrfind(x, params, t)

    idx = 2

    p_min = 0.0
    p_max = 1.0

    Δp = (p_max - p_min) / 50

    p_min += Δp
    p_max -= Δp

    return idx, Δp, p_max, p_min
end
```

```julia
seed = [ [q, p] for q in -1:0.2:1, p in -1:0.2:1]
```

```julia
E = 0.5
params = nothing

t0 = 0.0
tmax = 100.0

abstol = 1e-16
order = 20
```

```julia
crossings = PoincareMap(
    f!,
    bc!,
    g,
    H,
    nrfind,
    seed,
    E,
    params,
    tmax,
    abstol,
    order;
    maxsteps=1000,
    maxevents=100,
    eventorder=0,
    newtoniter=20,
    nrabstol=1e-14,
    Etol=1e-14,
)
```

```julia
@assert maximum(abs.(crossings[:, 1])) < 1e-12
```

```julia
@assert maximum(abs.(crossings[:, 2] .- 1.0)) < 1e-12
```

```julia
using PoincareMaps
using TaylorSeries

α = 0.1

function f!(dx, x, params, t)
    q1, q2, p1, p2 = x

    dx[1] = p1
    dx[2] = p2
    dx[3] = -q1 - α*q1^3
    dx[4] = -q2
end

function H(x, params, t)
    q1, q2, p1, p2 = x

    return (
        0.5*p1^2 +
        0.5*q1^2 +
        α*q1^4/4 +
        0.5*p2^2 +
        0.5*q2^2
    )
end

function bc!(x, params, t)
    nothing
end

function g(dx, x, params, t)
    return (dx[3][0] > 0, x[1])
end
```

```julia
function nrfind(x, params, t)

    idx = 4

    Δp2 = 0.01
    p2_min = -2.0
    p2_max =  2.0

    return idx, Δp2, p2_max, p2_min
end
```

```julia
seed = [
    [1.0, 0.0, 0.0, 0.0]
]
```

```julia
E = 1.0
params = nothing

tmax = 100.0
abstol = 1e-14
order = 20
```

```julia
crossings = PoincareMaps.PoincareMap(
    f!,
    bc!,
    g,
    H,
    nrfind,
    seed,
    E,
    params,
    tmax,
    abstol,
    order;
    maxsteps=5000,
    maxevents=100,
    eventorder=0,
    newtoniter=20,
    nrabstol=1e-14,
    Etol=1e-12,
)
```
